## Get data

- `12_29_2025/exp_0/run_0`: PMD on LunarLander-v3
- `12_29_2025/exp_0/run_1`: PPO on LunarLander-v3

```python
path = "/Users/calebju/Code/RL-general-action-state/logs"
# PMD on LunarLander-v3
df = pd.read_csv(os.path.join(path, "12_29_2025/exp_0/run_0/seed=0.csv"))
pmd_rwd_arr = df["episode rewards"].to_numpy()
pmd_cst_arr = -pmd_rwd_arr
pmd_len_arr = df["episode len"].to_numpy()

# PMD offline validation on LunarLander-v3
df = pd.read_csv(os.path.join(path, "12_29_2025/exp_0/run_0/offline_validation_seed=0.csv"))
pmd_V_arr = df["avg_V"].to_numpy()
pmd_V_lb_arr = df["avg_V_star_lb"].to_numpy()

# PPO on LunarLander-v3
df = pd.read_csv(os.path.join(path, "12_29_2025/exp_0/run_1/seed=0.csv"))
ppo_rwd_arr = df["episode rewards"].to_numpy()
ppo_cst_arr = -ppo_rwd_arr
ppo_len_arr = df["episode len"].to_numpy()

# PPO offline validation on LunarLander-v3
df = pd.read_csv(os.path.join(path, "12_29_2025/exp_0/run_1/offline_validation_seed=0.csv"))
ppo_V_arr = df["avg_V"].to_numpy()
ppo_V_lb_arr = df["avg_V_star_lb"].to_numpy()
```

## Plot progress

```python
plt.style.use('ggplot')
_, ax = plt.subplots(figsize=(5,4))

n_off_pmd = len(pmd_V_arr)
ax.plot(pmd_V_arr, label="PMD", linestyle="solid", color="red")
ax.plot(pmd_V_lb_arr, linestyle="dotted", color="red")
ax.plot(ppo_V_arr, label="PPO", linestyle="solid", color="gray")
ax.plot(ppo_V_lb_arr, linestyle="dotted", color="gray")

ax.set(
    title="Offline validation for LunarLander",
    xlabel="Episodes",
    ylabel="Upper and lower bounds of\nthe discounted objective",
)
ax.legend()
plt.tight_layout()
plt.savefig("lunarlander_validation.png", dpi=240)
```
